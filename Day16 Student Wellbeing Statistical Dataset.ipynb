```python
import pandas as pd
import numpy as np

from google.colab import files
uploaded = files.upload()

df = pd.read_csv("Day16_Student_Wellbeing_Survey.csv")

df.head()
```

```python
print(df.info())
print(df.describe())
print(df.isnull().sum())
```

```python
columns = ["Weekly_Study_Hours","Average_Sleep_Hours",
           "Daily_Screen_Time_Hours","Stress_Score",
           "Academic_Readiness_Score"]

for col in columns:
    print("\n", col)
    print("Mean:", df[col].mean())
    print("Median:", df[col].median())
    print("Mode:", df[col].mode()[0])
    print("Range:", df[col].max() - df[col].min())
    print("Variance:", df[col].var())
    print("Standard Deviation:", df[col].std())
```

```python
for col in columns:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1

    print("\n", col)
    print("Q1:", q1)
    print("Q3:", q3)
    print("IQR:", iqr)
```

```python
std_values = df[columns].std()
print(std_values)

print("Highest Variability:", std_values.idxmax())
```

```python
outlier_columns = ["Weekly_Study_Hours",
                   "Daily_Screen_Time_Hours",
                   "Commute_Time_Minutes",
                   "Monthly_Discretionary_Spending"]

for col in outlier_columns:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1

    lower = q1 - 1.5*iqr
    upper = q3 + 1.5*iqr

    outliers = df[(df[col] < lower) | (df[col] > upper)]

    print(col, "Outliers:", len(outliers))
```

```python
col = "Daily_Screen_Time_Hours"

q1 = df[col].quantile(0.25)
q3 = df[col].quantile(0.75)
iqr = q3 - q1

lower = q1 - 1.5*iqr
upper = q3 + 1.5*iqr

clean_df = df[(df[col] >= lower) & (df[col] <= upper)]

print("Mean Before:", df[col].mean())
print("Mean After:", clean_df[col].mean())

print("Median Before:", df[col].median())
print("Median After:", clean_df[col].median())
```

```python
A = len(df[df["Part_Time_Job"]=="Yes"]) / len(df)
B = len(df[df["Stress_Score"]>7]) / len(df)
C = len(df[df["Scholarship"]=="Yes"]) / len(df)
D = len(df[df["Exercise_Days_Per_Week"]>=3]) / len(df)

AB = len(df[(df["Part_Time_Job"]=="Yes") &
            (df["Stress_Score"]>7)]) / len(df)

AC = len(df[(df["Part_Time_Job"]=="Yes") &
            (df["Scholarship"]=="Yes")]) / len(df)

PA_given_B = AB / B
PB_given_A = AB / A

print("P(A):", A)
print("P(B):", B)
print("P(C):", C)
print("P(D):", D)
print("P(A∩B):", AB)
print("P(A∩C):", AC)
print("P(A|B):", PA_given_B)
print("P(B|A):", PB_given_A)
print("P(A)×P(B):", A*B)
```

```python
year1 = df[df["Year_of_Study"]==1]
year4 = df[df["Year_of_Study"]==4]

intersection = len(df[(df["Year_of_Study"]==1) &
                      (df["Year_of_Study"]==4)])

print("Intersection:", intersection)

if intersection == 0:
    print("Year 1 and Year 4 are mutually exclusive events.")
```

```python
mean = df["Academic_Readiness_Score"].mean()
std = df["Academic_Readiness_Score"].std()

df["Z_Score"] = (df["Academic_Readiness_Score"] - mean) / std

highest = df.loc[df["Academic_Readiness_Score"].idxmax()]
lowest = df.loc[df["Academic_Readiness_Score"].idxmin()]

print("Highest Student")
print(highest[["Student_ID","Academic_Readiness_Score","Z_Score"]])

print("Lowest Student")
print(lowest[["Student_ID","Academic_Readiness_Score","Z_Score"]])
```

```python
within1 = ((df["Z_Score"]>=-1) & (df["Z_Score"]<=1)).mean()*100
within2 = ((df["Z_Score"]>=-2) & (df["Z_Score"]<=2)).mean()*100
within3 = ((df["Z_Score"]>=-3) & (df["Z_Score"]<=3)).mean()*100

print("Within 1 SD:", within1,"%")
print("Within 2 SD:", within2,"%")
print("Within 3 SD:", within3,"%")
```
